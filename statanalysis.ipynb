*  Range is the simplest way to measure the
"spread" or "span" of your data. It tells you exactly how much space exists between your smallest and largest values.

*   Variance is a measure of how "spread out" your numbers are from the center (the average).
If the range tells you the distance between the two furthest people, variance tells you how much the whole crowd is scattered



*   Standard Deviation (SD) is the "average distance" that individual numbers in your dataset sit from the middle (the mean)












```python
import pandas as pd

data = {'marks1' : [40,50,60,70,80],
        'marks2' : [40,55,65,75,85]
        }
df = pd.DataFrame(data)

#range are calculated by substracting the minimum value form maximum value in dataset
range_marks1 = df['marks1'].max() - df['marks1'].min()
range_marks2 = df['marks2'].max() - df['marks2'].min()

print(f"The range for marks1 is : {range_marks1}")
print(f"The range for marks2 is : {range_marks2}")
```

```python
mean_marksss = df['marks1'].mean()
mean_marksss
```

```python
total_marks = df['marks2'].sum()
total_n = len(df['marks1'])
mean_mark = total_marks / total_n
print(f"The mean is :{mean_mark}")
# calculation x-sumofsquar
sum_of_marks_squared = 0

for x in df['marks2']:
  diff = x - mean_mark #40-60 yesari hudae sabai value minus hunxa
  squared = diff ** 2 #diff ko square hunxa
  sum_of_marks_squared += squared #sum

variance = sum_of_marks_squared / total_n - 1
variance


```

```python
def calclation(data_set):
  total_n = len(data_set)
  if total_n<= 1:
    return 0

  mean_data = sum(data_set)/total_n
  print(f"The mean data is: {mean_data}")

  # calculation x-sumofsquar
  sum_of_marks_squared = 0

  for x in data_set:
    diff = x - mean_mark #40-60 yesari hudae sabai value minus hunxa
    squared_diff = diff ** 2 #diff ko square hunxa
    sum_of_marks_squared += squared_diff #sum

  variance = sum_of_marks_squared / total_n -1 #calculating varience
  print(f"The variance is: {variance}")

  standard_deviation = variance ** 0.5 #stnadard deivation
  print(f"The standard deviation is: {standard_deviation}")
```

```python
data_set = df['marks1']
calclation(data_set)
```
